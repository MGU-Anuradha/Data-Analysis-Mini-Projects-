```python
import os  # Provides functions for interacting with the operating system
import shutil  # Provides functions for moving and copying files

# Set the path to the directory containing the files to be sorted
path = r'C:/Automatic File Sorter Project/'

# Get a list of all items (files and folders) in the specified directory
file_name = os.listdir(path)

# Define folder names for different file types
folder_names = ['csv files', 'image files', 'text files']

# Create folders for each file type if they don't already exist
for folder in folder_names:
    folder_path = os.path.join(path, folder)  # Join path and folder name
    if not os.path.exists(folder_path):  # Check if folder exists
        os.makedirs(folder_path)  # Create the folder

# Loop through each file in the directory
for file in file_name:
    # Create full path for the current file
    full_file_path = os.path.join(path, file)

    # Skip if it's a folder
    if os.path.isdir(full_file_path):
        continue

    # Move CSV files to 'csv files' folder
    if file.endswith('.csv') and not os.path.exists(os.path.join(path, 'csv files', file)):
        shutil.move(full_file_path, os.path.join(path, 'csv files', file))

    # Move PNG image files to 'image files' folder
    elif file.endswith('.png') and not os.path.exists(os.path.join(path, 'image files', file)):
        shutil.move(full_file_path, os.path.join(path, 'image files', file))

    # Move TXT text files to 'text files' folder
    elif file.endswith('.txt') and not os.path.exists(os.path.join(path, 'text files', file)):
        shutil.move(full_file_path, os.path.join(path, 'text files', file))
```
